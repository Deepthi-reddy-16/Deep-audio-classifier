```python
def extract_features(file_path, sr=44100, duration=5.0, n_mels=128, n_fft=2048, hop_length=512):
    """
    Extract mel spectrogram features from an audio file
    """
    try:
        # Load audio file with specified duration
        y, sr = librosa.load(file_path, sr=sr, duration=duration)

        # Add silent padding to short clips
        if len(y) < sr * duration:
            y = np.pad(y, (0, int(sr * duration) - len(y)), 'constant')

        # Extract mel spectrogram
        melspec = librosa.feature.melspectrogram(
            y=y, sr=sr, n_fft=n_fft, hop_length=hop_length, n_mels=n_mels)

        # Convert to logarithmic scale (dB)
        melspec_db = librosa.power_to_db(melspec, ref=np.max)

        return melspec_db

    except Exception as e:
        print(f"Error extracting features from {file_path}: {str(e)}")
        return None

def plot_features(file_path, title=None):
    """
    Plot waveform and mel spectrogram for an audio file
    """
    y, sr = librosa.load(file_path, sr=44100, duration=5.0)

    plt.figure(figsize=(14, 5))

    # Plot waveform
    plt.subplot(1, 2, 1)
    librosa.display.waveshow(y, sr=sr)
    plt.title('Waveform')

    # Plot mel spectrogram
    plt.subplot(1, 2, 2)
    melspec = librosa.feature.melspectrogram(y=y, sr=sr)
    melspec_db = librosa.power_to_db(melspec, ref=np.max)
    librosa.display.specshow(melspec_db, sr=sr, x_axis='time', y_axis='mel')
    plt.colorbar(format='%+2.0f dB')
    plt.title('Mel Spectrogram')

    if title:
        plt.suptitle(title)
    plt.tight_layout()
    plt.show()
```

```python
def time_shift(audio, shift_factor=0.1):
    """Shift audio in time by a random factor"""
    shift = int(len(audio) * shift_factor)
    direction = np.random.randint(0, 2)
    if direction == 1:
        # Shift right
        shifted_audio = np.pad(audio, (shift, 0), mode='constant')[:-shift]
    else:
        # Shift left
        shifted_audio = np.pad(audio, (0, shift), mode='constant')[shift:]
    return shifted_audio

def pitch_shift(audio, sr, n_steps=2):
    """Shift the pitch of audio by n_steps semitones"""
    return librosa.effects.pitch_shift(audio, sr=sr, n_steps=n_steps)

def add_noise(audio, noise_factor=0.005):
    """Add random noise to audio"""
    noise = np.random.randn(len(audio))
    return audio + noise_factor * noise

def augment_audio(audio, sr):
    """Apply random augmentations to audio"""
    augmentations = [
        lambda x: time_shift(x, shift_factor=np.random.uniform(0.05, 0.15)),
        lambda x: pitch_shift(x, sr, n_steps=np.random.uniform(-3, 3)),
        lambda x: add_noise(x, noise_factor=np.random.uniform(0.002, 0.01))
    ]

    # Randomly select augmentation(s)
    n_augment = np.random.randint(0, 4)  # 0 means no augmentation

    augmented_audio = audio.copy()
    if n_augment > 0:
        # Apply randomly selected augmentations
        augment_indices = np.random.choice(len(augmentations), n_augment, replace=False)
        for idx in augment_indices:
            augmented_audio = augmentations[idx](augmented_audio)

    return augmented_audio
```

```python
def prepare_dataset(metadata, with_augmentation=True, n_augmentations=1):
    """
    Extract features from audio files and prepare dataset
    with optional augmentation
    """
    features = []
    labels = []
    filenames = []

    # Process each audio file
    for idx, row in tqdm(metadata.iterrows(), total=len(metadata), desc="Processing audio files"):
        file_path = os.path.join(AUDIO_PATH, row['filename'])

        # Extract features from original audio
        mel_spec = extract_features(file_path)

        if mel_spec is not None:
            features.append(mel_spec)
            labels.append(row['target'])
            filenames.append(row['filename'])

            # Add augmented versions if requested
            if with_augmentation:
                y, sr = librosa.load(file_path, sr=44100, duration=5.0)

                for _ in range(n_augmentations):
                    # Apply random augmentations
                    augmented_y = augment_audio(y, sr)
                    augmented_mel_spec = librosa.feature.melspectrogram(
                        y=augmented_y, sr=sr)
                    augmented_mel_spec_db = librosa.power_to_db(augmented_mel_spec, ref=np.max)

                    features.append(augmented_mel_spec_db)
                    labels.append(row['target'])
                    filenames.append(row['filename'] + f"_aug_{_}")

    # Convert lists to numpy arrays
    features = np.array(features)
    labels = np.array(labels)

    # Reshape for CNN input: (samples, height, width, channels)
    features = np.expand_dims(features, axis=-1)

    return features, labels, filenames
```

```python
def build_efficientnet_model(input_shape, num_classes):
    """
    Build a model based on EfficientNetB0 with custom top layers
    """
    # Base model (pre-trained on ImageNet)
    # input_shape is modified here to have 3 channels
    modified_input_shape = (input_shape[0], input_shape[1], 3)
    base_model = EfficientNetB0(
        include_top=False,
        weights='imagenet',
        input_shape=modified_input_shape, #pass the modified shape
        pooling='avg'
    )

    # Repeat the mel spectrogram across 3 channels to simulate RGB
    inputs = Input(shape=input_shape)
    x = tf.keras.layers.Lambda(lambda x: tf.keras.backend.repeat_elements(x, rep=3, axis=-1))(inputs)

    # Freeze base model layers
    for layer in base_model.layers:
        layer.trainable = False

    # ... (rest of your build_efficientnet_model function) ...

    # Build the model
    inputs = Input(shape=input_shape)
    x = base_model(inputs)

    # Add custom top layers
    x = Dense(512, activation='relu')(x)
    x = Dropout(0.3)(x)
    x = Dense(256, activation='relu')(x)
    x = Dropout(0.3)(x)
    outputs = Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=inputs, outputs=outputs)

    # Compile the model
    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

def build_residual_cnn(input_shape, num_classes):
    """
    Build a custom CNN with residual connections
    """
    inputs = Input(shape=input_shape)

    # First convolutional block
    x = Conv2D(32, (3, 3), padding='same')(inputs)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = Conv2D(32, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = MaxPooling2D((2, 2))(x)
    x = Dropout(0.2)(x)

    # Residual block 1
    shortcut = Conv2D(64, (1, 1), padding='same')(x)
    x = Conv2D(64, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = Conv2D(64, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = Add()([x, shortcut])
    x = Activation('relu')(x)
    x = MaxPooling2D((2, 2))(x)
    x = Dropout(0.2)(x)

    # Residual block 2
    shortcut = Conv2D(128, (1, 1), padding='same')(x)
    x = Conv2D(128, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = Conv2D(128, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = Add()([x, shortcut])
    x = Activation('relu')(x)
    x = MaxPooling2D((2, 2))(x)
    x = Dropout(0.3)(x)

    # Residual block 3
    shortcut = Conv2D(256, (1, 1), padding='same')(x)
    x = Conv2D(256, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = Activation('relu')(x)
    x = Conv2D(256, (3, 3), padding='same')(x)
    x = BatchNormalization()(x)
    x = Add()([x, shortcut])
    x = Activation('relu')(x)
    x = MaxPooling2D((2, 2))(x)
    x = Dropout(0.4)(x)

    # Global pooling and output
    x = GlobalAveragePooling2D()(x)
    x = Dense(512, activation='relu')(x)
    x = Dropout(0.5)(x)
    outputs = Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=inputs, outputs=outputs)

    # Compile the model
    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
def get_fold_data(metadata, fold):
    """
    Split data into training and testing using the fold column in metadata
    """
    # Get test data for the current fold
    test_meta = metadata[metadata['fold'] == fold].copy()

    # Get training data for all other folds
    train_meta = metadata[metadata['fold'] != fold].copy()

    return train_meta, test_meta
```

```python
VALIDATION_FOLD = 5

# Get training and testing metadata
train_meta, test_meta = get_fold_data(metadata, VALIDATION_FOLD)

print(f"Training set size: {len(train_meta)} files")
print(f"Test set size: {len(test_meta)} files")

# Extract features with augmentation for training data
print("\nPreparing training data with augmentation...")
X_train, y_train, _ = prepare_dataset(train_meta, with_augmentation=True, n_augmentations=2)

# Extract features for test data (no augmentation)
print("\nPreparing test data...")
X_test, y_test, _ = prepare_dataset(test_meta, with_augmentation=False)

print(f"Training data shape: {X_train.shape}")
print(f"Testing data shape: {X_test.shape}")
```

```python
input_shape = X_train.shape[1:]
num_classes = len(metadata['target'].unique())

# Create a timestamp for model versioning
timestamp = datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
model_name = f"audio_classifier_{timestamp}"

# Build the model (choose one)
model = build_residual_cnn(input_shape, num_classes)
# Alternative: model = build_efficientnet_model(input_shape, num_classes)

model.summary()

# Set up callbacks
callbacks = [
    # Save the best model
    ModelCheckpoint(
        f"{model_name}_best.h5",
        monitor='val_accuracy',
        save_best_only=True,
        mode='max',
        verbose=1
    ),
    # Early stopping
    EarlyStopping(
        monitor='val_accuracy',
        patience=15,
        restore_best_weights=True,
        verbose=1
    ),
    # Reduce learning rate when plateauing
    ReduceLROnPlateau(
        monitor='val_accuracy',
        factor=0.5,
        patience=5,
        min_lr=1e-5,
        verbose=1
    )
]

# Train the model
history = model.fit(
    X_train, y_train,
    batch_size=32,
    epochs=40,  # We'll use early stopping
    validation_data=(X_test, y_test),
    callbacks=callbacks,
    verbose=1
)
```

```python
import gradio as gr
import numpy as np
import librosa
import librosa.display
import tensorflow as tf
import os

# Load your trained model
model = tf.keras.models.load_model("audio_classifier_20250414-070544_model1.h5")

# Define class labels (ESC-50)
classes = [
    'dog', 'rooster', 'pig', 'cow', 'frog', 'cat', 'hen', 'insects', 'sheep', 'crow',
    'rain', 'sea_waves', 'crackling_fire', 'crickets', 'chirping_birds', 'water_drops',
    'wind', 'pouring_water', 'toilet_flush', 'thunderstorm',
    'crying_baby', 'sneezing', 'clapping', 'breathing', 'coughing', 'footsteps',
    'laughing', 'brushing_teeth', 'snoring', 'drinking_sipping',
    'door_wood_knock', 'mouse_click', 'keyboard_typing', 'door_wood_creaks',
    'can_opening', 'washing_machine', 'vacuum_cleaner', 'clock_alarm',
    'clock_tick', 'glass_breaking', 'helicopter', 'chainsaw', 'siren',
    'car_horn', 'engine', 'train', 'church_bells', 'airplane', 'fireworks'
]

# Preprocessing: extract mel-spectrogram
def preprocess_audio(audio_path, sr=44100, n_mels=128, hop_length=512, n_fft=2048, duration=5):
    # Load audio file
    y, sr = librosa.load(audio_path, sr=sr, duration=duration)
    
    # Pad short clips
    if len(y) < sr * duration:
        y = np.pad(y, (0, sr * duration - len(y)))
        
    # Extract mel spectrogram
    mel = librosa.feature.melspectrogram(y=y, sr=sr, n_fft=n_fft, hop_length=hop_length, n_mels=n_mels)
    mel_db = librosa.power_to_db(mel, ref=np.max)
    
    # Add batch and channel dimensions for the model
    mel_db = mel_db[np.newaxis, ..., np.newaxis]
    return mel_db

# Prediction function
def predict_audio(file):
    mel_spec = preprocess_audio(file)
    preds = model.predict(mel_spec)[0]
    top_indices = preds.argsort()[-3:][::-1]
    top_classes = [(classes[i], float(preds[i])) for i in top_indices]
    return {label: score for label, score in top_classes}

# Gradio Interface
interface = gr.Interface(
    fn=predict_audio,
    inputs=gr.Audio(type="filepath", label="Upload Audio"),
    outputs=gr.Label(num_top_classes=3, label="Top Predictions"),
    title="Environmental Sound Classifier (ESC-50)",
    description="Upload a short audio clip (up to 5 seconds) and get the top predicted environmental sounds."
)

# Launch the interface directly (remove conditional)
interface.launch(share=True) # Removed the conditional and call launch directly
```
